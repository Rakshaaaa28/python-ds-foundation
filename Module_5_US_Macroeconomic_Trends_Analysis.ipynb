# Module 5 — Data Science Mini Project
## US Macroeconomic Trends Analysis

**Objective:** Explore a real-world public macroeconomic dataset and identify meaningful trends and relationships using Python, Pandas, NumPy, Matplotlib and Seaborn.

### Questions explored
1. How has real GDP changed over time?
2. How are real GDP and real consumption related?
3. How has unemployment changed across the period?
4. What periods show unusually high inflation?
5. How are major macroeconomic variables correlated?

**Dataset:** US Macroeconomic Data from the `statsmodels` public dataset collection.  
**Period:** 1959 Q1 to 2009 Q3.

## 1. Import Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option("display.max_columns", None)
sns.set_theme(style="whitegrid")
```

## 2. Load the Dataset

Upload **US_Macroeconomic_Data.csv** to Google Colab before running this cell.

```python
df = pd.read_csv("US_Macroeconomic_Data.csv")
df.head()
```

## 3. Understand the Dataset

```python
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

display(df.head())
display(df.info())
```

```python
df.describe().T
```

## 4. Data Cleaning and Preparation

The dataset is already structured, but we verify missing values and create a readable time label.

```python
print("Missing values by column:")
display(df.isnull().sum())

df["date"] = pd.PeriodIndex(
    year=df["year"].astype(int),
    quarter=df["quarter"].astype(int),
    freq="Q"
).to_timestamp()

df = df.sort_values("date").reset_index(drop=True)

print("Duplicate rows:", df.duplicated().sum())
```

## 5. Real GDP Trend

Real GDP is a useful indicator of the size and direction of economic activity. The chart below shows how it changed across the full period.

```python
plt.figure(figsize=(12, 5))
plt.plot(df["date"], df["realgdp"], linewidth=2)
plt.title("US Real GDP Over Time")
plt.xlabel("Year")
plt.ylabel("Real GDP")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## 6. Real Consumption Trend

```python
plt.figure(figsize=(12, 5))
plt.plot(df["date"], df["realcons"], linewidth=2)
plt.title("US Real Consumption Over Time")
plt.xlabel("Year")
plt.ylabel("Real Consumption")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## 7. GDP vs Consumption

This scatter plot helps examine whether higher economic output is associated with higher real consumption.

```python
plt.figure(figsize=(8, 6))
sns.scatterplot(data=df, x="realgdp", y="realcons")
plt.title("Real GDP vs Real Consumption")
plt.xlabel("Real GDP")
plt.ylabel("Real Consumption")
plt.tight_layout()
plt.show()

print("Correlation between real GDP and real consumption:",
      round(df["realgdp"].corr(df["realcons"]), 3))
```

## 8. Unemployment Over Time

```python
plt.figure(figsize=(12, 5))
plt.plot(df["date"], df["unemp"], linewidth=2)
plt.title("US Unemployment Rate Over Time")
plt.xlabel("Year")
plt.ylabel("Unemployment (%)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

print("Highest unemployment:")
display(df.loc[df["unemp"].idxmax(), ["date", "unemp"]])
```

## 9. Inflation Over Time

```python
plt.figure(figsize=(12, 5))
plt.plot(df["date"], df["infl"], linewidth=2)
plt.axhline(df["infl"].mean(), linestyle="--", linewidth=1.5, label="Average inflation")
plt.title("US Inflation Over Time")
plt.xlabel("Year")
plt.ylabel("Inflation")
plt.legend()
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

print("Average inflation:", round(df["infl"].mean(), 2))
print("Maximum inflation:", round(df["infl"].max(), 2))
```

## 10. Correlation Analysis

Correlation helps identify variables that move together. It does **not** prove causation.

```python
numeric_cols = [
    "realgdp", "realcons", "realinv", "realgovt",
    "realm1", "tbilrate", "unemp", "pop", "infl", "realint"
]

corr = df[numeric_cols].corr()

plt.figure(figsize=(11, 8))
sns.heatmap(corr, annot=True, fmt=".2f", cmap="coolwarm", center=0)
plt.title("Correlation Matrix of Major Macroeconomic Variables")
plt.tight_layout()
plt.show()
```

## 11. Key Summary Statistics

```python
summary = pd.DataFrame({
    "Metric": [
        "Highest real GDP",
        "Lowest real GDP",
        "Highest unemployment",
        "Lowest unemployment",
        "Highest inflation",
        "Lowest inflation"
    ],
    "Value": [
        df["realgdp"].max(),
        df["realgdp"].min(),
        df["unemp"].max(),
        df["unemp"].min(),
        df["infl"].max(),
        df["infl"].min()
    ]
})

summary
```

## 12. Findings and Insights

### Insight 1 — Long-term economic growth
Real GDP shows a strong long-term upward movement across the dataset, indicating substantial growth in the size of the US economy from the late 1950s to 2009.

### Insight 2 — Consumption follows economic activity
Real consumption also increases over the long term. The GDP-consumption correlation is strongly positive in this dataset, suggesting that periods of greater economic output generally coincide with higher consumption.

### Insight 3 — Unemployment is cyclical
Unemployment does not move smoothly upward or downward. It rises and falls over time, reflecting changing economic conditions and business cycles.

### Insight 4 — Inflation is uneven
Inflation varies considerably across quarters. Some periods show much higher inflation than the long-run average, demonstrating that price conditions can change sharply during certain economic periods.

### Insight 5 — Correlation is not causation
The correlation matrix identifies relationships between variables, but these relationships alone cannot establish that one variable causes another.
