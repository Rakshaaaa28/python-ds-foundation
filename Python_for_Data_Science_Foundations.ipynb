## 2. Lists


```python
scores = [78, 85, 92, 88, 76]
print('Scores:', scores)
print('Highest score:', max(scores))
print('Average score:', sum(scores) / len(scores))
```

## 3. Dictionaries


```python
student = {
    'name': 'Raksha',
    'track': 'Data Science',
    'completed_topics': 6
}

for key, value in student.items():
    print(f'{key}: {value}')
```

## 4. Loops


```python
for score in scores:
    if score >= 80:
        print(score, '-> Good')
    else:
        print(score, '-> Needs improvement')
```

## 5. Functions


```python
def calculate_average(values):
    return sum(values) / len(values)

average = calculate_average(scores)
print(f'Average score: {average:.2f}')
```

## 7. Data Science concepts and real-world applications

- **Data collection:** gathering useful data from databases, files, APIs, surveys, or other sources.
- **Data cleaning:** handling missing, duplicated, inconsistent, or incorrect data.
- **Exploratory Data Analysis (EDA):** examining data to find patterns, trends, and relationships.
- **Visualization:** communicating insights with charts and graphs.
- **Machine Learning:** building systems that learn patterns from data to make predictions or decisions.

### Examples
- Recommendation systems suggest products, movies, or music.
- Fraud detection can identify unusual transaction patterns.
- Customer analytics can help businesses understand customer behavior.
- Forecasting can support demand, sales, or resource planning.

## 8. Mini practice exercise

Calculate the average of the following monthly sales and classify each month as `Above Average` or `Below Average`.

```python
monthly_sales = {
    'January': 120,
    'February': 150,
    'March': 135,
    'April': 170,
    'May': 145
}

average_sales = sum(monthly_sales.values()) / len(monthly_sales)
print(f'Average monthly sales: {average_sales:.2f}\n')

for month, sales in monthly_sales.items():
    label = 'Above Average' if sales > average_sales else 'Below Average'
    print(f'{month}: {sales} -> {label}')
```
