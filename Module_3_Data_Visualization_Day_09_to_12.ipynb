## Setup
Import the libraries used throughout this module.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme()
df = pd.read_csv("student_performance.csv")
df.head()
```

# Day 9 — Matplotlib Basics

Create a bar chart, line chart, and pie chart.

```python
average_scores = df[["Math_Score", "Science_Score", "English_Score"]].mean()

plt.figure(figsize=(7, 4))
plt.bar(average_scores.index, average_scores.values)
plt.title("Average Score by Subject")
plt.xlabel("Subject")
plt.ylabel("Average Score")
plt.show()
```

```python
plt.figure(figsize=(9, 4))
plt.plot(df["Student_ID"], df["Math_Score"], marker="o")
plt.title("Math Scores Across Students")
plt.xlabel("Student ID")
plt.ylabel("Math Score")
plt.show()
```

```python
plt.figure(figsize=(6, 6))
plt.pie(average_scores.values, labels=["Math", "Science", "English"], autopct="%1.1f%%")
plt.title("Average Score Distribution by Subject")
plt.show()
```

# Day 10 — Histograms and Scatter Plots

```python
plt.figure(figsize=(7, 4))
plt.hist(df["Math_Score"], bins=6, edgecolor="black")
plt.title("Distribution of Math Scores")
plt.xlabel("Math Score")
plt.ylabel("Number of Students")
plt.show()
```

```python
plt.figure(figsize=(7, 4))
plt.scatter(df["Study_Hours"], df["Math_Score"])
plt.title("Study Hours vs Math Score")
plt.xlabel("Study Hours")
plt.ylabel("Math Score")
plt.show()
```

# Day 11 — Seaborn Visualizations

```python
plt.figure(figsize=(9, 4))
sns.barplot(data=df, x="Name", y="Math_Score")
plt.title("Math Score by Student")
plt.xlabel("Student")
plt.ylabel("Math Score")
plt.xticks(rotation=60)
plt.show()
```

```python
plt.figure(figsize=(7, 4))
sns.scatterplot(data=df, x="Study_Hours", y="Math_Score")
plt.title("Study Hours vs Math Score")
plt.show()
```

```python
score_data = df[["Math_Score", "Science_Score", "English_Score"]]
plt.figure(figsize=(7, 4))
sns.boxplot(data=score_data)
plt.title("Score Distribution by Subject")
plt.ylabel("Score")
plt.show()
```

```python
plt.figure(figsize=(8, 5))
correlation = df[["Study_Hours", "Attendance", "Math_Score", "Science_Score", "English_Score"]].corr()
sns.heatmap(correlation, annot=True, fmt=".2f")
plt.title("Correlation Between Study and Score Variables")
plt.show()
```

# Day 12 — Final Visualization & Insights

```python
plt.figure(figsize=(8, 5))
sns.regplot(data=df, x="Study_Hours", y="Math_Score")
plt.title("Study Hours and Math Score")
plt.xlabel("Study Hours")
plt.ylabel("Math Score")
plt.show()
```

```python
summary = {
    "Average Math Score": round(df["Math_Score"].mean(), 2),
    "Average Science Score": round(df["Science_Score"].mean(), 2),
    "Average English Score": round(df["English_Score"].mean(), 2),
    "Average Study Hours": round(df["Study_Hours"].mean(), 2),
    "Average Attendance": round(df["Attendance"].mean(), 2)
}
summary
```
