Practice project: exploring a student performance CSV using Pandas and NumPy.

```python
import pandas as pd
import numpy as np
```

## Load the CSV
Upload `student_performance.csv` to Colab using the Files panel, then run this cell.

```python
df = pd.read_csv('student_performance.csv')
df.head()
```

## Explore the dataset

```python
print('Shape:', df.shape)
print('Columns:', df.columns.tolist())
display(df.head())
display(df.tail())
```

```python
df.info()
print('\nMissing values:')
print(df.isnull().sum())
```

```python
df.describe()
```

## Basic analysis

```python
print('Average Math Score:', df['Math_Score'].mean())
print('Highest Math Score:', df['Math_Score'].max())
print('Lowest Math Score:', df['Math_Score'].min())
print('Average Study Hours:', df['Study_Hours'].mean())
print('Average Attendance:', df['Attendance'].mean())
```

## Filter and sort

```python
print('Students with Math score above 85:')
display(df[df['Math_Score'] > 85])

print('Top 5 Science scores:')
display(df.sort_values('Science_Score', ascending=False).head())
```

## NumPy practice

```python
science = df['Science_Score'].dropna().to_numpy()
print('Mean:', np.mean(science))
print('Median:', np.median(science))
print('Standard deviation:', np.std(science))
```
